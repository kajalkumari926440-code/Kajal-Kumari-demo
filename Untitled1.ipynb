```python

import os
import re
import datetime
import webbrowser

import speech_recognition as sr
import pyttsx3
import pywhatkit
from google import genai
from google.genai import types


# 1. Gemini API setup
API_KEY = os.getenv("GEMINI_API_KEY")

if not API_KEY:
    raise RuntimeError(
        "GEMINI_API_KEY nahi mili. Run Configuration check karo."
    )

client = genai.Client(api_key=API_KEY)

chat = client.chats.create(
    model="gemini-3.8-flash",
    config=types.GenerateContentConfig(
        system_instruction=(
            "You are Jarvis, Kajal's personal AI assistant. "
            "Reply in simple, natural Hinglish. "
            "Be friendly, helpful and accurate."
        ),
        temperature=0.4,
    ),
)


# 2. Voice engine
engine = pyttsx3.init("sapi5")
engine.setProperty("rate", 175)


def say(text):
    clean_text = re.sub(r"[*#_`~]", "", str(text))
    print(f"\nJarvis: {clean_text}")
    engine.say(clean_text)
    engine.runAndWait()


# 3. Listen to the user
def takeCommand():
    recognizer = sr.Recognizer()

    with sr.Microphone() as source:
        print("\nListening... Boliye!")
        recognizer.adjust_for_ambient_noise(source, duration=0.3)

        try:
            audio = recognizer.listen(
                source, timeout=5, phrase_time_limit=8
            )
        except sr.WaitTimeoutError:
            return ""

    try:
        query = recognizer.recognize_google(
            audio, language="en-IN"
        )
        print("You said:", query)
        return query.lower().strip()

    except sr.UnknownValueError:
        print("Awaaz samajh nahi aayi.")
        return ""

    except sr.RequestError:
        say("Voice recognition service connect nahi ho rahi.")
        return ""


# 4. Ask Gemini AI
def ask_gemini_brain(question):
    try:
        response = chat.send_message(question)
        if response.text:
            return response.text.strip()
        return "Mujhe jawab nahi mila. Dobara poochho."

    except Exception as error:
        print("Gemini error:", repr(error))
        return "AI connection mein problem hai. Baad mein try karo."


# 5. Main program
if __name__ == "__main__":
    say("Jarvis online hai! Hello Kajal, bataiye main kya karoon?")

    shutdown_pending = False

    while True:
        query = takeCommand()

        if not query:
            continue

        # Shutdown confirmation
        if shutdown_pending:
            if query in ("yes", "confirm shutdown", "haan"):
                say("Computer das second mein band hoga.")
                os.system("shutdown /s /t 10")
                break
            else:
                shutdown_pending = False
                say("Shutdown cancel kar diya.")
                continue

        # Exit Jarvis
        if query in ("exit jarvis", "close jarvis", "jarvis band karo"):
            say("Theek hai Kajal, Jarvis band ho raha hai.")
            break

        # Open YouTube
        elif "youtube" in query and any(
            word in query for word in ("open", "kholo", "launch")
        ):
            say("YouTube open kar raha hoon.")
            webbrowser.open("https://www.youtube.com")

        # Play a song
        elif query.startswith("play "):
            song = query[5:].replace("on youtube", "").strip()
            if song:
                say(f"YouTube par {song} chala raha hoon.")
                pywhatkit.playonyt(song)
            else:
                say("Kya play karna hai, Kajal?")

        # Open Google
        elif "google" in query and any(
            word in query for word in ("open", "kholo", "launch")
        ):
            say("Google open kar raha hoon.")
            webbrowser.open("https://www.google.com")

        # Tell time
        elif "time" in query or "time batao" in query:
            current_time = datetime.datetime.now().strftime("%I:%M %p")
            say(f"Kajal, abhi time {current_time} hai.")

        # Ask AI for other questions
        else:
            answer = ask_gemini_brain(query)
            say(answer)
```
